```python
import os

# 「experiment_data」という名前のフォルダ（ディレクトリ）を新しく作ります
os.makedirs("experiment_data", exist_ok=True)

# for文を使って、10個のファイルを自動生成する
for i in range(1, 11):
    # ファイル名を「data_01.txt」「data_02.txt」のように連番にする
    file_name = f"experiment_data/data_{i:02d}.txt"

    # 空のファイルを作成して書き込む
    with open(file_name, "w") as f:
        f.write(f"This is experiment data number {i}\n")

print("10個のファイルを自動で作成しました！")
```
