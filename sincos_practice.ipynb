```python
import matplotlib.pyplot as plt
import numpy as np

# ① 横軸を「0 から 2π までの間」で100等分する
x = np.linspace(0, 4 * np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

# グラフのサイズを設定
plt.figure(figsize=(16, 4))

plt.plot(x, y_sin, color="blue", label="sinx")
plt.plot(x, y_cos, color="red", label="cosx")

# ② 横軸の目盛りを「0, π/2, π, 3π/2, 2π」の位置に固定する
positions = [0, np.pi, 2 * np.pi, 3 * np.pi, 4 * np.pi]
labels = ["0", "π", "2π", "3π", "4π"]     # 画面に表示する文字
plt.xticks(positions, labels)                  # 位置と文字をセットで登録！

plt.grid(True)
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("sinx and cosx")
plt.show()
```
