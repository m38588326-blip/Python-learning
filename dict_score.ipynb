```python
score_dict = {"イトウ":85,"タナカ":50,"ヤマダ":90}

for name, score in score_dict.items():
  if score >= 60:
    print(f"{name}さん：合格（{score}点）")
  else:
    print(f"{name}さん：不合格（{score}点）")
```
