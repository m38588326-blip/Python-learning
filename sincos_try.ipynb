```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(20, 4))

x = np.linspace(0, 2*np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, color="blue", label="sinx")
plt.plot(x, y_cos, color="red", label="cosx")

positions = [0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi]
labels = ["0", "π/2", "π", "3π/2", "2π"]     # 画面に表示する文字
plt.xticks(positions, labels)
plt.yticks(np.arange(-1., 1.1, 0.5))

plt.xlabel("x")
plt.ylabel("y")
plt.title("sinx and cosx")
plt.legend()
plt.grid(True)
plt.show()
```
