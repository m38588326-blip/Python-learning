```python
import matplotlib.pyplot as plt
import numpy as np

# ① -5から5までの間に、均等な間隔で50個の数字（データ）を作ります（NumPyの技）
x = np.linspace(-5, 5, 50)

# ② 数学の「二次関数」の計算を一瞬で行います
y = x ** 2

# ③ グラフを描画します（Matplotlibの技）
plt.plot(x, y, marker="o", color="blue")  # 点を青い線で結ぶ
plt.title("My First Graph")                # グラフのタイトル
plt.xlabel("X-Axis")                       # 横軸のラベル
plt.ylabel("Y-Axis")                       # 縦軸のラベル
plt.grid(True)                             # 背景に方眼紙のようなマス目を出す

# ④ 画面に表示する
plt.show()
```
